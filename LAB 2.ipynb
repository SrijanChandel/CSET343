```python

# LAB ASSIGNMENT - 2

# DATA CLEANING AND ENRICHMENT
# HEART FAILURE CLINICAL RECORDS DATASET




# 1. IMPORT REQUIRED LIBRARIES


import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler



# 2. DATASET ACQUISITION AND LOADING

file_path = r"C:\Personal AI & ML PROJECTS\AI IN HEALTHCARE\heart_failure_clinical_records_dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.\n")

print("First 5 rows of the dataset:")
display(df.head())



# 3. SUMMARIZE DATASET


print("\n================ DATASET SUMMARY ================\n")

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nSummary Statistics:")
display(df.describe())



# 4. VISUALIZE MISSING VALUES

missing_data = df.isnull().sum().to_frame(name="Missing Values")

plt.figure(figsize=(7, 8))

sns.heatmap(
    missing_data,
    annot=True,
    fmt="d",
    cmap="YlOrRd",
    cbar=False,
    linewidths=0.5
)

plt.title("Missing Values by Feature")
plt.xlabel("")
plt.ylabel("Features")

plt.tight_layout()
plt.show()




missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

print("\nMissing Value Summary:")
display(missing_summary)



# 5. PLOT DISTRIBUTIONS OF NUMERICAL FEATURES


continuous_features = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium",
    "time"
]


# AGE HISTOGRAM 

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="age",
    bins=20,
    kde=True
)

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# EJECTION FRACTION

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="ejection_fraction",
    bins=15,
    kde=True
)

plt.title("Distribution of Ejection Fraction")
plt.xlabel("Ejection Fraction")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# SERUM CREATININE BOXPLOT 

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df,
    x="serum_creatinine"
)

plt.title("Serum Creatinine Before Cleaning")
plt.xlabel("Serum Creatinine")

plt.tight_layout()
plt.show()


# PLATELETS BOXPLOT 

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df,
    x="platelets"
)

plt.title("Platelets Before Cleaning")
plt.xlabel("Platelets")

plt.tight_layout()
plt.show()


# ALL NUMERICAL FEATURES 

df[continuous_features].hist(
    figsize=(15, 10),
    bins=20
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16
)

plt.tight_layout()
plt.show()



# 6. IMPUTE MISSING VALUES


df_clean = df.copy()


# Numerical variables
numerical_features = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium",
    "time"
]


# Binary / categorical variables
categorical_features = [
    "anaemia",
    "diabetes",
    "high_blood_pressure",
    "sex",
    "smoking"
]


print("\nMissing Values Before Imputation:")
print(df_clean.isnull().sum())


# ---------------- NUMERICAL IMPUTATION ----------------


for column in numerical_features:

    if df_clean[column].isnull().sum() > 0:

        median_value = df_clean[column].median()

        df_clean[column] = df_clean[column].fillna(
            median_value
        )

        print(
            f"{column}: Missing values filled using median = {median_value}"
        )


# CATEGORICAL IMPUTATION

for column in categorical_features:

    if df_clean[column].isnull().sum() > 0:

        mode_value = df_clean[column].mode()[0]

        df_clean[column] = df_clean[column].fillna(
            mode_value
        )

        print(
            f"{column}: Missing values filled using mode = {mode_value}"
        )


print("\nMissing Values After Imputation:")
print(df_clean.isnull().sum())

print(
    "\nTotal Missing Values After Imputation:",
    df_clean.isnull().sum().sum()
)



# STORE ORIGINAL CLEANING BASELINE

df_before_cleaning = df_clean.copy()



# 7. REMOVE OUTLIERS USING IQR METHOD

outlier_columns = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium"
]


def remove_outliers_iqr(data, columns):

    cleaned_data = data.copy()

    outlier_report = []

    for column in columns:

        Q1 = cleaned_data[column].quantile(0.25)
        Q3 = cleaned_data[column].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        rows_before = len(cleaned_data)

        cleaned_data = cleaned_data[
            (cleaned_data[column] >= lower_bound) &
            (cleaned_data[column] <= upper_bound)
        ]

        rows_after = len(cleaned_data)

        removed = rows_before - rows_after

        outlier_report.append({
            "Feature": column,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Rows Removed": removed
        })

    return cleaned_data, pd.DataFrame(outlier_report)


print("\nRows Before Outlier Removal:", len(df_clean))

df_clean, outlier_report = remove_outliers_iqr(
    df_clean,
    outlier_columns
)

print("Rows After Outlier Removal:", len(df_clean))

print(
    "Total Rows Removed:",
    len(df_before_cleaning) - len(df_clean)
)


print("\nOutlier Removal Report:")
display(outlier_report)




df_clean = df_clean.reset_index(drop=True)



# 8. CORRECT DATA INCONSISTENCIES


print("\n================ CONSISTENCY CHECK ================\n")


# ---------------- INVALID AGE ----------------

invalid_age = (df_clean["age"] <= 0).sum()

print("Invalid Age Values:", invalid_age)


# INVALID EJECTION FRACTION 

invalid_ef = (
    (df_clean["ejection_fraction"] < 0) |
    (df_clean["ejection_fraction"] > 100)
).sum()

print("Invalid Ejection Fraction Values:", invalid_ef)


# INVALID SERUM CREATININE 

invalid_creatinine = (
    df_clean["serum_creatinine"] < 0
).sum()

print(
    "Invalid Serum Creatinine Values:",
    invalid_creatinine
)


#INVALID PLATELETS 

invalid_platelets = (
    df_clean["platelets"] < 0
).sum()

print(
    "Invalid Platelet Values:",
    invalid_platelets
)


# REMOVE INVALID RECORDS

df_clean = df_clean[
    df_clean["age"] > 0
]

df_clean = df_clean[
    (df_clean["ejection_fraction"] >= 0) &
    (df_clean["ejection_fraction"] <= 100)
]

df_clean = df_clean[
    df_clean["serum_creatinine"] >= 0
]

df_clean = df_clean[
    df_clean["platelets"] >= 0
]


#  CHECK BINARY VARIABLES 

binary_columns = [
    "anaemia",
    "diabetes",
    "high_blood_pressure",
    "sex",
    "smoking",
    "DEATH_EVENT"
]

print("\nUnique Values in Binary Variables:")

for column in binary_columns:

    print(
        column,
        "->",
        sorted(df_clean[column].unique())
    )




for column in binary_columns:

    df_clean = df_clean[
        df_clean[column].isin([0, 1])
    ]


df_clean = df_clean.reset_index(drop=True)


print(
    "\nRows After Consistency Validation:",
    len(df_clean)
)


# 9. FEATURE ENGINEERING




# 9A. CREATE AGE GROUP

df_clean["age_group"] = pd.cut(
    df_clean["age"],
    bins=[
        -np.inf,
        40,
        60,
        np.inf
    ],
    labels=[
        "<40",
        "40-60",
        ">60"
    ],
    right=True
)


print("\nAge Group Distribution:")
print(df_clean["age_group"].value_counts())


print("\nExample Age Group Values:")

display(
    df_clean[
        ["age", "age_group"]
    ].head(10)
)


# 9B. CREATE RISK SCORE

risk_scaler = MinMaxScaler()

risk_variables = risk_scaler.fit_transform(
    df_clean[
        [
            "ejection_fraction",
            "serum_creatinine"
        ]
    ]
)


df_clean["ef_norm"] = risk_variables[:, 0]

df_clean["creatinine_norm"] = risk_variables[:, 1]




df_clean["ef_risk"] = (
    1 - df_clean["ef_norm"]
)


# Risk score

df_clean["risk_score"] = (
    df_clean["ef_risk"] *
    df_clean["creatinine_norm"]
)


print("\nEngineered Risk Score:")

display(
    df_clean[
        [
            "ejection_fraction",
            "serum_creatinine",
            "ef_norm",
            "creatinine_norm",
            "risk_score"
        ]
    ].head(10)
)


print(
    "\nRisk Score Minimum:",
    df_clean["risk_score"].min()
)

print(
    "Risk Score Maximum:",
    df_clean["risk_score"].max()
)



# 10. ENCODE CATEGORICAL VARIABLES



df_clean = pd.get_dummies(
    df_clean,
    columns=["age_group"],
    prefix="age_group",
    dtype=int
)


print("\nDataset After Encoding:")

display(df_clean.head())


print("\nEncoded Age Group Columns:")

age_group_columns = [
    column
    for column in df_clean.columns
    if column.startswith("age_group")
]

print(age_group_columns)



# SAVE DATASET BEFORE NORMALIZATION

df_after_cleaning_before_scaling = df_clean.copy()



# 11. NORMALIZE NUMERICAL FEATURES USING MINMAXSCALER


features_to_scale = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium",
    "time"
]


scaler = MinMaxScaler()


df_clean[features_to_scale] = scaler.fit_transform(
    df_clean[features_to_scale]
)


print("\nNormalized Features:")

display(
    df_clean[
        features_to_scale
    ].head()
)


# Check minimum and maximum after normalization

normalization_check = pd.DataFrame({
    "Minimum": df_clean[features_to_scale].min(),
    "Maximum": df_clean[features_to_scale].max()
})


print("\nNormalization Validation:")
display(normalization_check)



# 12. COMPUTE SUMMARY STATISTICS

comparison_features = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium",
    "time"
]


before_stats = (
    df_before_cleaning[
        comparison_features
    ]
    .agg(
        [
            "mean",
            "median",
            "std"
        ]
    )
    .T
)


after_stats = (
    df_after_cleaning_before_scaling[
        comparison_features
    ]
    .agg(
        [
            "mean",
            "median",
            "std"
        ]
    )
    .T
)


before_stats.columns = [
    "Mean Before",
    "Median Before",
    "Std Before"
]


after_stats.columns = [
    "Mean After",
    "Median After",
    "Std After"
]


comparison_stats = pd.concat(
    [
        before_stats,
        after_stats
    ],
    axis=1
)


print(
    "\nSummary Statistics Before and After Cleaning:"
)

display(
    comparison_stats.round(3)
)



# 13. VALIDATE CLEANING USING VISUALIZATIONS



# AGE AFTER CLEANING

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_after_cleaning_before_scaling,
    x="age",
    bins=20,
    kde=True
)

plt.title("Age Distribution After Cleaning")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


#  EJECTION FRACTION AFTER CLEANING 

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_after_cleaning_before_scaling,
    x="ejection_fraction",
    bins=15,
    kde=True
)

plt.title(
    "Ejection Fraction Distribution After Cleaning"
)

plt.xlabel("Ejection Fraction")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


#  SERUM CREATININE AFTER CLEANING 

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df_after_cleaning_before_scaling,
    x="serum_creatinine"
)

plt.title(
    "Serum Creatinine After Cleaning"
)

plt.xlabel("Serum Creatinine")

plt.tight_layout()
plt.show()


#  PLATELETS AFTER CLEANING 

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df_after_cleaning_before_scaling,
    x="platelets"
)

plt.title(
    "Platelets After Cleaning"
)

plt.xlabel("Platelets")

plt.tight_layout()
plt.show()


#  ALL NUMERICAL FEATURES 

df_after_cleaning_before_scaling[
    comparison_features
].hist(
    figsize=(15, 10),
    bins=20
)

plt.suptitle(
    "Distribution of Numerical Features After Cleaning",
    fontsize=16
)

plt.tight_layout()
plt.show()



# 14. FINAL MISSING VALUE CHECK


print(
    "\n================ FINAL VALIDATION ================\n"
)


print("Missing Values in Final Dataset:")

print(
    df_clean.isnull().sum()
)


final_missing_values = (
    df_clean.isnull().sum().sum()
)


print(
    "\nTotal Missing Values Remaining:",
    final_missing_values
)


if final_missing_values == 0:

    print(
        "\nSUCCESS: No missing values remain in the cleaned dataset."
    )

else:

    print(
        "\nWARNING: Some missing values still remain."
    )


# Additional validation

assert final_missing_values == 0


print(
    "\nFinal Dataset Shape:",
    df_clean.shape
)


print(
    "\nFinal Dataset Preview:"
)

display(df_clean.head())


# SAVE FINAL CLEANED DATASET


output_path = r"C:\Personal AI & ML PROJECTS\AI IN HEALTHCARE\heart_failure_cleaned.csv"


df_clean.to_csv(
    output_path,
    index=False
)


print(
    "\nCleaned dataset saved successfully at:"
)

print(output_path)
```
